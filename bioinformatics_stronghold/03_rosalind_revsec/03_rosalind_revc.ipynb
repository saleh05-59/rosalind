```python
# take the original strand
original_dna = input("please Enter the DNA String: ")

# reverse the original strand
original_dna = original_dna[::-1]

# make it complementary (T-A)
original_dna = original_dna.replace('T', 'X')
original_dna = original_dna.replace('A', 'Y')
original_dna = original_dna.replace('X', 'A')
original_dna = original_dna.replace('Y', 'T')

# make it complementary (C-G)

original_dna = original_dna.replace('C', 'X')
original_dna = original_dna.replace('G', 'Y')
original_dna = original_dna.replace('X', 'G')
original_dna = original_dna.replace('Y', 'C')

# print the result
print(original_dna)
```

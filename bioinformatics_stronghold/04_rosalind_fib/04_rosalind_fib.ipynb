```python
# getting the data from the file
given_data = 0
with open('rosalind_fib.txt', 'r') as input_file:
    given_data = input_file.read()
given_data = given_data.split()
```

```python
# defining the function ig
def fibonacci_prime(cycles, rabbit_pair):
    total_rabbits = [1, 1]
    for n in range(cycles-2):
        current_round = len(total_rabbits) + 1
        f_1 = total_rabbits[current_round-3] # correct digit for 0 based counting of the len function
        f_2 = total_rabbits[current_round-2]
        total_rabbits.append((f_1 * (rabbit_pair)) + f_2)

    return total_rabbits[-1]
```

```python
fibonacci_prime(int(given_data[0]), int(given_data[1]))
```
